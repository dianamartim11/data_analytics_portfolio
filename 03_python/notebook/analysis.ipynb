```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()

P = '../../data/processed'
orders    = pd.read_csv(f'{P}/orders_clean.csv', parse_dates=['order_date'])
products  = pd.read_csv(f'{P}/products_clean.csv')
customers = pd.read_csv(f'{P}/customers_clean.csv')
orders.head()
```

## 1. Build the analysis table
Join orders to products and customers, then add `revenue` and `profit`.

```python
df = (orders
      .merge(products, on='product_id', how='left')
      .merge(customers[['customer_id','county','gender','age']], on='customer_id', how='left'))
df['revenue'] = df['quantity'] * df['unit_price']
df['profit']  = (df['unit_price'] - df['cost_price']) * df['quantity']
df.describe()
```
